<a id="eda-missing-values-train-vs-test" name="eda-missing-values-train-vs-test"></a>

## 6. Missing Values: Train vs. Test

```python
missing_compare = pd.DataFrame({
    "train_missing_pct": (train.drop(columns="addicted_label").isna().mean() * 100).round(3),
    "test_missing_pct": (test.isna().mean() * 100).round(3),
})

missing_compare["difference"] = (
    missing_compare["test_missing_pct"]
    - missing_compare["train_missing_pct"]
).round(3)

display(
    missing_compare.sort_values(
        "train_missing_pct",
        ascending=False
    )
)
```

## 7. Summary

| Area | Main finding |
|---|---|
| Dataset | 691,369 training rows, 296,302 test rows, and 12 predictive features |
| Target | 70.9% positive vs. 29.1% negative |
| Missingness | 4.2%–19.4% across all predictive features; the largest train–test gap is 3.4 percentage points |

### Modelling implications

- The target imbalance makes stratified validation important, while the moderate train–test missingness differences should be monitored during evaluation.
- With 9 numerical features, 3 categorical features, and widespread missing values, CatBoost is a natural baseline because it can preserve numerical `NaN`s and handle categorical variables directly.

### Next direction

Build a simple CatBoost baseline first, using explicit missing-value labels for categorical columns and leaving numerical missing values to the model.
